```python
target_table=f"{catalog_name}.{gold_schema}.dim_races"
#target_table=f"{catalog_name}.{gold_schema}.dim_races"
```

```python
circuits_df=spark.table(f"{catalog_name}.{bronze_schema}.circuits")
races_df=spark.table(f"{catalog_name}.{bronze_schema}.races")
```

```python
from pyspark.sql.functions import col

dim_races_df=(
        races_df
        .join
        (circuits_df, races_df.circuitId==circuits_df.circuitid, "inner")
        .select(col("season"),
            col("round"),
            col("raceName").alias("race_name"),
            col("date").alias("race_date"),
            col("circuitName").alias("circuit_name"),
            col("locality"),
            col("country")
        )
 
)
```

```python
(
    dim_races_df
        .write
        .format("delta")
        .mode("overwrite")
        .saveAsTable(target_table)
)
```
