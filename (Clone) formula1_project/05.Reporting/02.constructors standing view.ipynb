```python
%sql
create or replace view formula1.gold.v_constructor_standing as
with constructors_session_summary 
as (select  r.season,
c.constructor_id,
c.constructors_name,
c.nationality,
count(*) as race_starts,
count_if(r.is_win) as num_of_wins,
sum(r.points) as total_points,
count_if(r.is_podium) as num_of_podiums
 from formula1.gold.fact_results r
join formula1.gold.dim_constructors c
on r.constructor_id = c.constructor_id
group by
r.season,
c.constructor_id,
c.constructors_name,
c.nationality)
select season,constructor_id,constructors_name,nationality,race_starts,num_of_wins,num_of_podiums,
total_points,
rank() over(partition by season
order by total_points desc, num_of_wins desc) as standing from constructors_session_summary;
```
