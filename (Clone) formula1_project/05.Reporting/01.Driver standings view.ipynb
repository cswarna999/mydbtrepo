```python
%sql
create or replace view formula1.gold.v_driver_standing as
with driver_session_summary 
as (select r.season,
d.driver_id,
d.driver_name,
d.nationality,
count(*) as race_starts,
count_if(r.is_win) as num_of_wins,
count_if(r.is_podium) as num_of_podiums,
sum(r.points) as total_points from 
formula1.gold.fact_results r
join formula1.gold.dim_drivers d
on r.driver_id=d.driver_id
GROUP BY r.season,
d.driver_id,
d.driver_name,
d.nationality)
select 
season,driver_id,driver_name,nationality,race_starts,num_of_wins,num_of_podiums,
total_points,
rank() over(partition by season
order by total_points desc, num_of_wins desc) as standing from driver_session_summary ;
```
