```python
bronze_table=f"{catalog_name}.{bronze_schema}.races"
silver_table=f"{catalog_name}.{silver_schema}.races"
```

```python
races_df=spark.read.table(bronze_table)
```

```python
from pyspark.sql import functions as F
```

```python
races_selected_df=(races_df.select(F.col("season"),
                   F.col("round"),
                   F.col("raceName"),
                   F.col("date"),
                   F.col("circuitId"),
                   F.col("ingestion_timestamp"),
                   F.col("source_file")
                   )
)
```

```python
races_renamed_df=(races_selected_df
                         .withColumnsRenamed({"circuitId":"circuit_id","date":"race_date","raceName":"race_name"})
)
```

```python
races_distinct_df=races_renamed_df.dropDuplicates(["season", "round"])
```

```python
races_final_df=(races_distinct_df
                   .withColumn("race_name",F.initcap(F.col("race_name")))
)
```

```python
(
    races_final_df
       .write
       .format("delta")
       .mode("overwrite")
       .saveAsTable(silver_table)
)
```
